```python
# 15) Elabore um código que utilize um laço while para fazer uma contagem regressiva,
# começando em 10 e indo até 0. Quando a contagem terminar, imprima
# a palavra "Decolar!"

contador = 10
while contador >= 0:
  if contador == 0:
    print("Decolar!")
  else:
    print(contador)
  contador -= 1
```

```python
# 18) Elabore uma função chamada calcula_desconto que receba dois parâmetros:
# o valor de um produto e a porcentagem de desconto. A função deve retornar o
# valor final do produto com o desconto aplicado. Teste a função e imprima o
# resultado.

def calcula_desconto():
  valor_produto = float(input("Valor do produto: "))
  porcentagem_de_desconto = float(input("Porcentagem de desconto: "))

  desconto = valor_produto * (porcentagem_de_desconto / 100)

  valor_final = valor_produto - desconto

  print(f"Desconto: R$ {desconto:.2f}")
  print(f"Valor final do produto: R$ {valor_final:.2f}")
calcula_desconto()
```

```python
# 19) Elabore uma função chamada classifica_nota que receba uma nota de 0 a 10.
# A função deve retornar "Aprovado" se a nota for maior ou igual a 7, "Recuperação"
# se estiver entre 5 e 6.9, e "Reprovado" se for menor que 5. Chame a função
# passando uma nota e imprima o retorno.

def classifica_nota():
  nota = float(input("Digite uma nota: "))
  if nota >= 7:
    print(f"Nota: {nota}. \nSituação: Aprovado.")
  elif nota >= 5:
    print(f"Nota: {nota}. \nSituação: Recuperação.")
  else:
    print(f"Nota: {nota}. \nSituação: Reprovado.")

  return nota

classifica_nota()
```

```python
# 20) Elabore um código que receba uma lista de números inteiros mistos
# (positivos, negativos e zeros). Utilize um laço for com estruturas condicionais
# if/elif/else para contabilizar quantos números são positivos, quantos são
# negativos e quantos são exatamente zero. Ao final, imprima esse resumo.

positivos = 0
negativos = 0
zeros = 0

lista = [-1, -2, 0, 0, 0, 0, 4, 3, 5, 8, 10, 12, -4, -3]
for numero in lista:
  if numero > 0:
    positivos += 1
  elif numero < 0:
    negativos += 1
  else:
    zeros += 1

print(f"Quantidade de número positivos: {positivos}")
print(f"Quantidade de números negativos: {negativos}")
print(f"Quantidade de zeros: {zeros}")
```
