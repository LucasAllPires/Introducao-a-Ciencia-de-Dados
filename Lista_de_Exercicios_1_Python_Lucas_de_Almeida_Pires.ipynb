```python
# 1) Elabore um código que imprima a mensagem *"Olá terráqueos!!"* na tela.

print("Olá terráqueos")
```

```python
# 2) Elabore um código que peça um número e então mostre a mensagem: "O número informado foi [número]."
numero = float(input("Digite um número: "))
print(f"O número informado foi {numero}.")
```

```python
# 3) Elabore um código que peça dois números e imprima a soma.
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))
print(f"A soma dos dois números é igual a {numero1 + numero2}")
```

```python
3# 4) Elabore um código que peça as 4 notas bimestrais e mostre a média.
nota1 = float(input("Nota 1: "))
nota2 = float(input("Nota 2: "))
nota3 = float(input("Nota 3: "))
nota4 = float(input("Nota 4: "))
media = (nota1 + nota2 + nota3 + nota4)/4
print(f"A sua média do bimestre é igual a {media:.2f}")
```

```python
# 5) Elabore um código que converta metros para centímetros.
valor_metro = float(input("Digite a medida em metros: "))
valor_cm = valor_metro * 100
print(f"A sua medida em cm é igual a {valor_cm}.")
```

```python
# 6) Elabore um código que peça o raio de um círculo, calcule e mostre sua área.
raio = float(input("Digite o raio do círculo: "))
area = 3.14 * raio**2
print(f"A área desse círculo é igual a {area:.2f}")
```

```python
# 7) Elabore um código que calcule a área de um quadrado, em seguida mostre o dobro desta área para o usuário.
lado_quadrado = float(input("Digite o lado do quadrado: "))
area_quadrado1 = lado_quadrado ** 2
print(f"O dobro da área desse quadrado é {area_quadrado1*2:.2f}")
```

```python
# 8) Elabore um código que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês. Calcule e mostre o total do seu salário no referido mês em uma mensagem amigável.
dinheiro_hora = float(input("Olá! Vamos descobrir quanto você ganha por mês! Para isso, responda: quanto você ganha por hora trabalhada? "))
horas_trabalhadas_mes = int(input("Quantas horas trabalha por mês? "))
salario = dinheiro_hora * horas_trabalhadas_mes
print(f"Parabéns! Seu salário total no mês é de R${salario:.2f}")
```

```python
# 9) Elabore um código que peça a temperatura em graus Fahrenheit, transforme e mostre a temperatura em graus Celsius. ** Onde C = 5 * ((F-32) / 9)**.
temperatura_f = float(input("Temperatura em Fahrenheit: "))
temperatura_c = 5 * ((temperatura_f - 32)/9)
print(f"A temperatura em graus celsius é igual a {temperatura_c:.2f}")
```

```python
# 10) Elabore um código que peça a temperatura em graus Celsius, transforme e mostre em graus Fahrenheit.
temperatura_c = float(input("Temperatura em Celsius: "))
temperatura_f = 1.8 * temperatura_c + 32
print(f"A temperatura em Celsius é igual a {temperatura_f:.2f}.")
```

```python
# 11) Elabore um código que peça 2 números inteiros e um número real. Calcule e mostre: o produto do dobro do primeiro com metade do segundo, a soma do triplo do primeiro com o terceiro e o terceiro elevado ao cubo.
numero1 = int(input("Digite o primeiro número (inteiro): "))
numero2 = int(input("Digite o segundo número (inteiro): "))
numero3 = float(input("Digite o terceiro número (real): "))

# a) O produto do dobro do primeiro com metade do segundo
produto = (2 * numero1) * (numero2/2)
print(f"O produto do dobro do primeiro com metade do segundo: {produto}")

# b) A soma do triplo do primeiro com o terceiro
soma = numero1 * 3 + numero3
print(f"A soma do triplo do primeiro com o terceiro é {soma}")

# c) O terceiro elevado ao cubo
print(f"O terceiro número elevado ao cubo é igual a {numero3 ** 3:.2f}")
```

```python
# 12) Elabore um código que receba os dados de entrada a altura de uma pessoa, e informe qual seria o peso ideal, usando a seguinte fórmula: (72.7*altura) - 58
altura = float(input("Vamos descobrir qual é o seu peso ideal. Digite sua altura: "))
peso_ideal = (72.7*altura) - 58
print(f"O seu peso ideal é {peso_ideal:.2f}.")
```
